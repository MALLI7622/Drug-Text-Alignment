```python
from chembl_webresource_client.new_client import new_client
import pandas as pd
from itertools import islice
from tqdm import tqdm

molecule = new_client.molecule

# 1) Unique IDs only
ids = (
    df["molecule_chembl_id"]
    .dropna()
    .astype(str)
    .unique()
    .tolist()
)

print(len(ids))
```

```python


# 2) Helper: chunk a list
def chunked(iterable, size):
    it = iter(iterable)
    for first in it:
        yield [first, *list(islice(it, size-1))]

# 3) Pull only the fields we need, in batches (e.g., 200 per call)
id_to_smiles = {}
batches = list(chunked(ids, 200))  # precompute batches for tqdm length

for batch in tqdm(batches, desc="Fetching SMILES in batches"):
    q = molecule.filter(molecule_chembl_id__in=",".join(batch)).only([
        "molecule_chembl_id",
        "molecule_structures"
    ])
    for rec in q:
        mid = rec.get("molecule_chembl_id")
        smi = (rec.get("molecule_structures") or {}).get("canonical_smiles")
        if mid and smi:
            id_to_smiles[mid] = smi

# 4) Map back to your dataframe (vectorized, no per-row HTTP)
df["smiles"] = df["molecule_chembl_id"].map(id_to_smiles)
```

```python
# ==== 3) Helper: mean-pool last hidden state with attention mask ====
def mean_pool(last_hidden_state, attention_mask):
    # last_hidden_state: [B, T, H], attention_mask: [B, T]
    mask = attention_mask.unsqueeze(-1).type_as(last_hidden_state)
    summed = (last_hidden_state * mask).sum(dim=1)
    counts = mask.sum(dim=1).clamp(min=1e-9)
    return summed / counts
```

```python
@torch.no_grad()
def encode_texts(texts, tok, model, max_length=128, batch_size=32):
    reps = []
    for i in tqdm(range(0, len(texts), batch_size), desc="Encoding texts"):
        batch = texts[i:i+batch_size]
        enc = tok(batch, padding=True, truncation=True, max_length=max_length, return_tensors="pt").to(device)
        out = model(**enc)
        emb = mean_pool(out.last_hidden_state, enc["attention_mask"])
        reps.append(emb.detach().cpu())
    return torch.cat(reps, dim=0)
```

```python
@torch.no_grad()
def encode_smiles(smiles, tok, model, max_length=128, batch_size=32):
    reps = []
    for i in tqdm(range(0, len(smiles), batch_size), desc="Encoding SMILES"):
        batch = smiles[i:i+batch_size]
        enc = tok(batch, padding=True, truncation=True, max_length=128, return_tensors="pt").to(device)
        out = model(**enc)
        emb = mean_pool(out.last_hidden_state, enc["attention_mask"])
        reps.append(emb.detach().cpu())
    return torch.cat(reps, dim=0)
```

```python
import torch, torch.nn as nn, torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import math, random

torch.manual_seed(0); np.random.seed(0); random.seed(0)

# Cache the frozen CLS embeddings from step 1 (or 2)
Zt = E_text.clone()      # [N, Ht]
Zs = E_smiles.clone()    # [N, Hs]

class PairDS(Dataset):
    def __init__(self, Zt, Zs):
        self.Zt = Zt; self.Zs = Zs
    def __len__(self): return len(self.Zt)
    def __getitem__(self, i): return self.Zt[i], self.Zs[i]

ds = PairDS(Zt, Zs)
dl = DataLoader(ds, batch_size=256, shuffle=True, drop_last=True)

d_text = Zt.shape[1]; d_smiles = Zs.shape[1]; d_sh = 256

proj_text   = nn.Linear(d_text, d_sh).to(device)
proj_smiles = nn.Linear(d_smiles, d_sh).to(device)

opt = optim.AdamW(list(proj_text.parameters())+list(proj_smiles.parameters()), lr=1e-3, weight_decay=1e-4)
T = 0.07  # temperature

def contrastive_step(bt, bs):
    # bt, bs: [B, d]
    bt = F.normalize(proj_text(bt.to(device)), dim=1)
    bs = F.normalize(proj_smiles(bs.to(device)), dim=1)
    logits = (bt @ bs.T) / T     # [B, B]
    labels = torch.arange(bt.size(0), device=bt.device)
    loss_i = F.cross_entropy(logits, labels)
    loss_j = F.cross_entropy(logits.T, labels)
    loss = (loss_i + loss_j) / 2
    return loss

epochs = 100  # ~5–10 minutes on Colab; reduce to 3 if needed
for ep in range(1, epochs+1):
    proj_text.train(); proj_smiles.train()
    running = 0.0; steps = 0
    for bt, bs in dl:
        opt.zero_grad()
        loss = contrastive_step(bt, bs)
        loss.backward()
        opt.step()
        running += loss.item(); steps += 1
    if ep % 10 == 0:
        print(f"Epoch {ep}/{epochs}  loss={running/steps:.4f}")

# Evaluate with learned heads
@torch.no_grad()
def eval_with_heads(Zt, Zs):
    Bt = F.normalize(proj_text(Zt.to(device)), dim=1).cpu()
    Bs = F.normalize(proj_smiles(Zs.to(device)), dim=1).cpu()
    S = Bt @ Bs.T
    gt = np.arange(S.shape[0])
    top1 = S.argmax(dim=1).numpy()
    recall1 = (top1 == gt).mean()
    ranks = []
    for i in range(S.shape[0]):
        order = torch.argsort(S[i], descending=True).numpy()
        rank_pos = int(np.where(order == i)[0][0]) + 1
        ranks.append(1.0 / rank_pos)
    mrr = float(np.mean(ranks))
    return recall1, mrr

rec1, mrr = eval_with_heads(Zt, Zs)
print(f"[After tiny contrastive bridge] Recall@1: {rec1:.3f} | MRR: {mrr:.3f}")
```

```python
# === Heatmap for the learned-bridge model ===
import torch, numpy as np, seaborn as sns, matplotlib.pyplot as plt
import torch.nn.functional as F

@torch.no_grad()
def similarity_with_heads(Zt, Zs, proj_text, proj_smiles):
    # Zt: [N, d_text] CLS text embeddings (frozen)
    # Zs: [N, d_smiles] CLS smiles embeddings (frozen)
    Bt = F.normalize(proj_text(Zt.to(device)), dim=1).cpu()   # [N, d_shared]
    Bs = F.normalize(proj_smiles(Zs.to(device)), dim=1).cpu() # [N, d_shared]
    S  = Bt @ Bs.T                                            # [N, N]
    return S

# Build similarity matrix with the learned heads
S_heads = similarity_with_heads(Zt, Zs, proj_text, proj_smiles)

# Small heatmap (first K×K block for readability)
K = min(40, S_heads.shape[0])
plt.figure(figsize=(7,5))
sns.heatmap(S_heads[:K, :K], vmin=-1, vmax=1, cmap="viridis")
plt.title("Cosine similarity after learned bridge (first K)")
plt.xlabel("Molecule index")
plt.ylabel("Text index")
plt.tight_layout()
plt.show()

# Optional: compute metrics from this S (should match your eval_with_heads)
gt = np.arange(S_heads.shape[0])
order = torch.argsort(S_heads, dim=1, descending=True).numpy()
top1 = order[:, 0]
recall_at_1 = (top1 == gt).mean()

ranks = []
for i in range(S_heads.shape[0]):
    # where is the true pair j=i in the sorted order?
    rank_pos = int(np.where(order[i] == i)[0][0]) + 1
    ranks.append(1.0 / rank_pos)
mrr = float(np.mean(ranks))
print(f"[Bridge heatmap eval] Recall@1: {recall_at_1:.3f} | MRR: {mrr:.3f}")
```

```python
import numpy as np, torch, torch.nn.functional as F
from rdkit import Chem
import copy
from rdkit.Chem import AllChem
from torch.utils.data import DataLoader, Dataset
import torch.nn as nn, torch.optim as optim
import pandas as pd, random
import warnings; warnings.filterwarnings("ignore")

device = "cuda" if torch.cuda.is_available() else "cpu"
df = pd.read_csv("chembl_mechanisms_enriched.csv").dropna(subset=["smiles","text_rich"])
df = df[df["text_rich"].str.len() > 20].reset_index(drop=True)

# 1) Compute ECFP4 (2048 bits)
def ecfp4_bitvect(smi, nbits=2048, radius=2):
    m = Chem.MolFromSmiles(smi)
    if m is None: return np.zeros(nbits, dtype=np.float32)
    bv = AllChem.GetMorganFingerprintAsBitVect(m, radius, nBits=nbits)
    arr = np.zeros((nbits,), dtype=np.int8)
    Chem.DataStructs.ConvertToNumpyArray(bv, arr)
    return arr.astype(np.float32)

X_mol = np.stack([ecfp4_bitvect(s) for s in df["smiles"].tolist()], axis=0)  # [N,2048]

# 2) Get text embeddings (reuse your best text encoder from earlier steps)
#    Assume you already have a function cls_encode(...) and loaded text_tok/text_model
from transformers import AutoTokenizer, AutoModel
text_model_name = "pritamdeka/S-Biomed-Roberta-snli-multinli-stsb"  # or your best
text_tok = AutoTokenizer.from_pretrained(text_model_name)
text_model = AutoModel.from_pretrained(text_model_name).to(device).eval()

@torch.no_grad()
def cls_encode(texts, tok, model, max_length=96, bs=64):
    outs = []
    for i in range(0, len(texts), bs):
        batch = texts[i:i+bs]
        enc = tok(batch, padding=True, truncation=True, max_length=max_length, return_tensors="pt").to(device)
        out = model(**enc).last_hidden_state[:,0,:]
        outs.append(out.detach().cpu())
    return torch.cat(outs, dim=0)

Z_text = cls_encode(df["text_rich"].tolist(), text_tok, text_model)  # [N, Ht]

# 3) Contrastive training: frozen text, learn only mol projector
class PairDS(Dataset):
    def __init__(self, Zt, Xm):
        self.Zt = torch.tensor(Zt, dtype=torch.float32)
        self.Xm = torch.tensor(Xm, dtype=torch.float32)
    def __len__(self): return len(self.Zt)
    def __getitem__(self, i): return self.Zt[i], self.Xm[i]

ds = PairDS(Z_text.numpy(), X_mol)
dl = DataLoader(ds, batch_size=512, shuffle=True, drop_last=True)

d_text = Z_text.shape[1]; d_fp = X_mol.shape[1]; d_sh = 256
proj_text = nn.Linear(d_text, d_sh).to(device)     # learn small head (or fix to identity via nn.Identity())
proj_mol  = nn.Linear(d_fp,   d_sh).to(device)

proj_text_0 = copy.deepcopy(proj_text)
proj_mol_0  = copy.deepcopy(proj_mol)

opt = optim.AdamW(list(proj_text.parameters())+list(proj_mol.parameters()), lr=1e-3, weight_decay=1e-4)
T = 0.07

def step(bt, bm):
    bt = F.normalize(proj_text(bt.to(device)), dim=1)
    bm = F.normalize(proj_mol (bm.to(device)), dim=1)
    logits = bt @ bm.T / T
    labels = torch.arange(bt.size(0), device=bt.device)
    loss = (F.cross_entropy(logits, labels) + F.cross_entropy(logits.T, labels)) / 2
    return loss

for ep in range(1, 101):  # start with 3 epochs (fast)
    tot = n = 0
    for bt, bm in dl:
        opt.zero_grad()
        loss = step(bt, bm)
        loss.backward()
        opt.step()
        tot += loss.item(); n += 1
    print(f"ECFP4 bridge epoch {ep} loss={tot/n:.4f}")

# 4) Evaluate Recall@1/MRR
@torch.no_grad()
def eval_recall_mrr(Zt, Xm):
    Bt = F.normalize(proj_text(torch.tensor(Zt, dtype=torch.float32).to(device)), dim=1).cpu()
    Bm = F.normalize(proj_mol (torch.tensor(Xm, dtype=torch.float32).to(device)), dim=1).cpu()
    S = Bt @ Bm.T
    gt = np.arange(S.shape[0])
    top1 = S.argmax(dim=1).numpy()
    recall1 = (top1 == gt).mean()
    ranks = []
    for i in range(S.shape[0]):
        order = torch.argsort(S[i], descending=True).numpy()
        r = int(np.where(order == i)[0][0]) + 1
        ranks.append(1.0/r)
    return recall1, float(np.mean(ranks))

rec1, mrr = eval_recall_mrr(Z_text.numpy(), X_mol)
print(f"[ECFP4 bridge] Recall@1: {rec1:.3f} | MRR: {mrr:.3f}")
```

```python


import os, math, random, numpy as np, pandas as pd, torch, torch.nn as nn, torch.nn.functional as F
from tqdm import tqdm
from rdkit import Chem
from rdkit.Chem import AllChem
from rdkit.Chem.Scaffolds import MurckoScaffold
from transformers import AutoTokenizer, AutoModel
from torch.utils.data import Dataset, DataLoader

device = "cuda" if torch.cuda.is_available() else "cpu"
print("Device:", device)

# =============================
# Config
# =============================
CSV_PATH = "chembl_mechanisms_enriched.csv"   # must have: smiles, text_rich, target_chembl_id (plus any others)
TEXT_MODEL = "pritamdeka/S-Biomed-Roberta-snli-multinli-stsb"  # swap if needed
MAX_TXT_LEN = 96
BATCH_SIZE = 512
EPOCHS = 21                  # 6-10 works well on Colab
LR = 1e-3
WEIGHT_DECAY = 1e-4
TEMP = 0.07                 # InfoNCE temperature
SHARED_D = 256              # projection size
ECFP_BITS = 2048
ECFP_RADIUS = 2
SEED = 0

# Evaluation toggles
REMOVE_DRUG_NAME = False    # leakage control: if your text_rich contains "Drug: {pref_name}", set True to strip it
MIN_GROUP_SIZE = 3          # for grouped Recall@1
TEST_ONLY_TOPK = None       # e.g., 500 to speed eval; None = all

# =============================
# Utils
# =============================
def set_seed(seed=0):
    random.seed(seed); np.random.seed(seed); torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)

def ecfp4_bitvect(smi, nbits=2048, radius=2):
    m = Chem.MolFromSmiles(smi)
    if m is None:
        return np.zeros(nbits, dtype=np.float32)
    bv = AllChem.GetMorganFingerprintAsBitVect(m, radius, nBits=nbits)
    arr = np.zeros((nbits,), dtype=np.int8)
    Chem.DataStructs.ConvertToNumpyArray(bv, arr)
    return arr.astype(np.float32)

def murcko(sm):
    m = Chem.MolFromSmiles(sm)
    if m is None: return None
    core = MurckoScaffold.GetScaffoldForMol(m)
    return Chem.MolToSmiles(core) if core else None

@torch.no_grad()
def cls_encode(texts, tok, model, max_length=96, bs=64):
    outs = []
    for i in tqdm(range(0, len(texts), bs), desc="Encode text (CLS)"):
        batch = texts[i:i+bs]
        enc = tok(batch, padding=True, truncation=True, max_length=max_length, return_tensors="pt").to(device)
        out = model(**enc).last_hidden_state[:,0,:]  # CLS token
        outs.append(out.detach().cpu())
    return torch.cat(outs, dim=0)

def build_group_index(series):
    """Map group label -> list of indices"""
    groups = series.fillna("UNK").astype(str).values
    idx_by_group = {}
    for i, g in enumerate(groups):
        idx_by_group.setdefault(g, []).append(i)
    return idx_by_group

def grouped_recall_at1(S_np, meta_df, group_col="target_chembl_id", min_group=3):
    """Recall@1 computed within each group, then micro-averaged."""
    assert S_np.shape[0] == S_np.shape[1] == len(meta_df)
    idx_by_group = build_group_index(meta_df[group_col])
    hits = total = 0
    for g, idxs in idx_by_group.items():
        if len(idxs) < min_group:
            continue
        sub = S_np[np.ix_(idxs, idxs)]
        top1 = sub.argmax(axis=1)
        hits += (top1 == np.arange(len(idxs))).sum()
        total += len(idxs)
    return hits/total if total else 0.0

# =============================
# 1) Load & prepare data
# =============================
set_seed(SEED)
df = pd.read_csv(CSV_PATH)
need_cols = {"smiles","text_rich","target_chembl_id"}
missing = need_cols - set(df.columns)
if missing:
    raise ValueError(f"CSV missing required columns: {missing}")

df = df.dropna(subset=["smiles","text_rich"]).copy()

if REMOVE_DRUG_NAME:
    # crude removal of "Drug: X." pattern if present in text_rich
    df["text_rich"] = df["text_rich"].str.replace(r"Drug:\s*[^.]+\. ?", "", regex=True).str.strip()

# compute Murcko scaffolds and split
print("Computing Murcko scaffolds...")
df["scaffold"] = df["smiles"].apply(murcko)
df = df.dropna(subset=["scaffold"]).reset_index(drop=True)
scaffolds = df["scaffold"].unique()
np.random.seed(SEED); np.random.shuffle(scaffolds)
cut = int(0.9*len(scaffolds))
train_scafs = set(scaffolds[:cut])
test_scafs  = set(scaffolds[cut:])

train_df = df[df["scaffold"].isin(train_scafs)].reset_index(drop=True)
test_df  = df[df["scaffold"].isin(test_scafs)].reset_index(drop=True)
print(f"Split sizes -> train: {len(train_df)} | test: {len(test_df)}")

# Optional: limit test size for speed
if TEST_ONLY_TOPK is not None and len(test_df) > TEST_ONLY_TOPK:
    test_df = test_df.sample(TEST_ONLY_TOPK, random_state=SEED).reset_index(drop=True)
    print(f"Test subset for speed: {len(test_df)}")

# =============================
# 2) Build features: ECFP4 (mol) + Text CLS (frozen)
# =============================
print("Computing ECFP4 fingerprints...")
X_mol_train = np.stack([ecfp4_bitvect(s, ECFP_BITS, ECFP_RADIUS) for s in tqdm(train_df["smiles"].tolist())])
X_mol_test  = np.stack([ecfp4_bitvect(s, ECFP_BITS, ECFP_RADIUS) for s in tqdm(test_df["smiles"].tolist())])

print("Encoding text embeddings...")
text_tok = AutoTokenizer.from_pretrained(TEXT_MODEL, use_fast=True)
text_model = AutoModel.from_pretrained(TEXT_MODEL).to(device).eval()
Z_text_train = cls_encode(train_df["text_rich"].tolist(), text_tok, text_model, max_length=MAX_TXT_LEN, bs=64)
Z_text_test  = cls_encode(test_df["text_rich"].tolist(),  text_tok, text_model, max_length=MAX_TXT_LEN, bs=64)

# =============================
# 3) Dataset/Loader for training bridge
# =============================
class PairDataset(Dataset):
    def __init__(self, Z_text, X_mol, group_labels):
        self.Zt = torch.tensor(Z_text, dtype=torch.float32)
        self.Xm = torch.tensor(X_mol,  dtype=torch.float32)
        self.grp = np.array(group_labels).astype(str)
    def __len__(self): return len(self.Zt)
    def __getitem__(self, i):
        return self.Zt[i], self.Xm[i], self.grp[i], i

train_ds = PairDataset(Z_text_train.numpy(), X_mol_train, train_df["target_chembl_id"].tolist())
train_dl = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True, drop_last=True)

# =============================
# 4) Thin contrastive bridge with hard-neg weighting + margin loss
# =============================
d_text = Z_text_train.shape[1]
d_fp   = X_mol_train.shape[1]

proj_text = nn.Linear(d_text, SHARED_D).to(device)
proj_mol  = nn.Linear(d_fp,   SHARED_D).to(device)

proj_text_0 = copy.deepcopy(proj_text)
proj_mol_0  = copy.deepcopy(proj_mol)

opt = torch.optim.AdamW(list(proj_text.parameters())+list(proj_mol.parameters()), lr=LR, weight_decay=WEIGHT_DECAY)

def build_same_target_weights(grps):
    """Return weight matrix W (B,B): w>1 for same-target negatives, 1 otherwise; diagonal left to CE."""
    B = len(grps)
    W = torch.ones((B,B), device=device)
    for i in range(B):
        for j in range(B):
            if i != j and grps[i] == grps[j]:
                W[i,j] = 2.0  # up-weight same-target impostors
    return W

def hardest_same_target_neg(logits, groups):
    """
    For each row i, find the best same-target negative j!=i.
    Return tensor of size [B] with the max logit among same-target negatives; if none, return -inf.
    """
    B = logits.size(0)
    hard = torch.full((B,), float("-inf"), device=logits.device)
    for i in range(B):
        mask = torch.tensor([ (groups[i]==groups[j] and j!=i) for j in range(B) ], device=logits.device)
        if mask.any():
            hard[i] = logits[i][mask].max()
    return hard

def contrastive_loss_with_weights(bt, bm, groups, T=0.07, margin=0.1):
    # Normalize
    bt = F.normalize(bt, dim=1)
    bm = F.normalize(bm, dim=1)
    logits = (bt @ bm.T) / T                      # [B,B]
    labels = torch.arange(bt.size(0), device=bt.device)

    # Weighted InfoNCE (row + col) with same-target emphasis
    W = build_same_target_weights(groups)         # [B,B]
    # Row-wise CE
    ce_row = F.cross_entropy(logits, labels, reduction='none')   # [B]
    # Column-wise CE
    ce_col = F.cross_entropy(logits.T, labels, reduction='none') # [B]
    # Weight by average same-target pressure (normalize to keep scales reasonable)
    # We multiply per-example CE by (1 + average weight on its negatives - 1) = average(W[i,neg])
    with torch.no_grad():
        B = logits.size(0)
        avg_w_row = (W.sum(dim=1) - torch.diag(W)) / (B-1)   # [B]
        avg_w_col = (W.sum(dim=0) - torch.diag(W)) / (B-1)   # [B]
    loss_row = (ce_row * avg_w_row).mean()
    loss_col = (ce_col * avg_w_col).mean()
    loss_nce = 0.5*(loss_row + loss_col)

    # Margin loss vs. hardest same-target impostor
    pos = logits.diag()                                      # [B]
    hard_same = hardest_same_target_neg(logits, groups)      # [B] (−inf if none)
    # hinge only where a same-target negative exists
    mask = torch.isfinite(hard_same)
    margin_loss = torch.clamp(margin - (pos[mask] - hard_same[mask]), min=0).mean() if mask.any() else torch.tensor(0.0, device=logits.device)

    return loss_nce + margin_loss, logits
```

```python
EPOCHS = 101
# Train
set_seed(SEED)
for ep in range(1, EPOCHS+1):
    proj_text.train(); proj_mol.train()
    running, n = 0.0, 0
    for Zt_b, Xm_b, grp_b, idx_b in tqdm(train_dl, desc=f"Epoch {ep}/{EPOCHS}"):
        Zt_b = Zt_b.to(device); Xm_b = Xm_b.to(device)
        Bt = proj_text(Zt_b);   Bm = proj_mol(Xm_b)

        loss, _ = contrastive_loss_with_weights(Bt, Bm, grp_b, T=TEMP, margin=0.1)
        opt.zero_grad(); loss.backward(); opt.step()

        running += loss.item(); n += 1
    print(f"Epoch {ep}: loss={running/max(n,1):.4f}")

# =============================
# 5) Evaluate on TEST split
# =============================
@torch.no_grad()
def embed_side(Z_text, X_mol):
    Bt = F.normalize(proj_text(torch.tensor(Z_text, dtype=torch.float32).to(device)), dim=1).cpu()
    Bm = F.normalize(proj_mol (torch.tensor(X_mol,  dtype=torch.float32).to(device)), dim=1).cpu()
    return Bt, Bm

Bt_test, Bm_test = embed_side(Z_text_test.numpy(), X_mol_test)
S = (Bt_test @ Bm_test.T)  # [N,N]
S_np = S.numpy()

# Global metrics
gt = np.arange(S_np.shape[0])
top1 = S_np.argmax(axis=1)
recall1 = (top1 == gt).mean()

ranks = []
for i in range(S_np.shape[0]):
    order = np.argsort(-S_np[i])
    rank_pos = int(np.where(order == i)[0][0]) + 1
    ranks.append(1.0 / rank_pos)
mrr = float(np.mean(ranks))

print(f"\n=== TEST (Scaffold split) ===")
print(f"Global Recall@1: {recall1:.3f}")
print(f"Global MRR:      {mrr:.3f}")

# Grouped (within-target) Recall@1
gR1 = grouped_recall_at1(S_np, test_df, group_col="target_chembl_id", min_group=MIN_GROUP_SIZE)
print(f"Grouped Recall@1 (within-target): {gR1:.3f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
import torch
import torch.nn.functional as F

def tsne_2d(X, n_components=2, perplexity=30, random_state=0):
    """PCA→t-SNE for stability."""
    X = np.asarray(X, dtype=np.float32)
    X50 = PCA(n_components=min(50, X.shape[1])).fit_transform(X)
    X2  = TSNE(
        n_components=n_components,
        perplexity=perplexity,
        learning_rate=200,
        init="pca",
        random_state=random_state,
    ).fit_transform(X50)
    return X2

import numpy as np
import matplotlib.pyplot as plt

import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

def plot_tsne_points(X2, labels, title, cmap="tab20", markers=None, modalities=None):
    """
    X2      : (N, 2) t-SNE coordinates
    labels  : categorical labels (can be CHEMBL IDs, NaN, etc.)
    modalities : optional array ['text', 'molecule', ...] for marker types
    """
    plt.figure(figsize=(6, 6))

    # 1) Make labels a 1D pandas Series
    labels = pd.Series(labels)

    # 2) Convert everything to string (so CHEMBL IDs, NaN etc are unified)
    labels = labels.astype("string")

    # 3) Factorize -> integer IDs per unique label (NO sorting, NO type comparison)
    label_ids, uniques = pd.factorize(labels, sort=False)

    if modalities is None:
        # single-modality scatter
        plt.scatter(X2[:, 0], X2[:, 1],
                    c=label_ids, cmap=cmap, s=10)
    else:
        modalities = np.asarray(modalities)
        uniq_mods = np.unique(modalities)

        for m in uniq_mods:
            mask = modalities == m
            marker = markers.get(m, "o") if markers else "o"
            plt.scatter(
                X2[mask, 0], X2[mask, 1],
                c=label_ids[mask], cmap=cmap,
                s=12, marker=marker, alpha=0.8,
                label=m,
            )
        plt.legend(markerscale=1.8, fontsize=8, loc="best")

    plt.title(title)
    plt.xlabel("t-SNE feature 1")
    plt.ylabel("t-SNE feature 2")
    plt.tight_layout()
    plt.savefig(f"{title.lower().replace(' ', '_')}.pdf")
    plt.savefig(f"{title.lower().replace(' ', '_')}.png")
    plt.show()


@torch.no_grad()
def project_into_shared_space(Zt_np, Xm_np, pt, pm, device=device):
    Zt = torch.tensor(Zt_np, dtype=torch.float32).to(device)
    Xm = torch.tensor(Xm_np, dtype=torch.float32).to(device)

    Bt = F.normalize(pt(Zt), dim=1).cpu().numpy()  # text → 256D
    Bm = F.normalize(pm(Xm), dim=1).cpu().numpy()  # mol  → 256D
    return Bt, Bm
```

```python
import numpy as np

def recall_at1(sim):
    gt = np.arange(sim.shape[0])
    return (sim.argmax(axis=1) == gt).mean()

def grouped_r1(sim, df, col="target_chembl_id", min_group=3):
    groups = df[col].fillna("UNK").astype(str).values
    idx = {}
    for i,g in enumerate(groups): idx.setdefault(g, []).append(i)
    hits=total=0
    for g,ix in idx.items():
        if len(ix)<min_group: continue
        sub = sim[np.ix_(ix,ix)]
        hits += (sub.argmax(axis=1) == np.arange(len(ix))).sum()
        total += len(ix)
    return hits/total if total else 0.0

def bootstrap_ci(sim_np, meta_df, metric_fn, B=1000, seed=0):
    rng = np.random.default_rng(seed)
    N = sim_np.shape[0]
    vals=[]
    for _ in range(B):
        samp = rng.choice(N, size=N, replace=True)
        S = sim_np[np.ix_(samp, samp)]
        vals.append(metric_fn(S, meta_df.iloc[samp].reset_index(drop=True)))
    a,b = np.percentile(vals, [2.5,97.5])
    return float(np.mean(vals)), float(a), float(b)

# After you compute S_np on test split:
g_r1_mean, g_lo, g_hi = bootstrap_ci(S_np, test_df, lambda S,df: recall_at1(S))
grp_mean, grp_lo, grp_hi = bootstrap_ci(S_np, test_df, lambda S,df: grouped_r1(S, df))
print(f"Global R@1: {g_r1_mean:.3f} [{g_lo:.3f},{g_hi:.3f}]")
print(f"Grouped R@1: {grp_mean:.3f} [{grp_lo:.3f},{grp_hi:.3f}]")
```

```python
# Plotting toolkit for your workshop figures
# This cell defines three plots:
# 1) Within-target similarity heatmap
# 2) Top-k retrieval curves (CMC) for Global and Within-target
# 3) Ablation bar chart for Grouped R@1
#
# How it works:
# - It first tries to use variables already in memory: S_np (NxN similarity), test_df (pandas DataFrame), res (ablation DataFrame).
# - If those aren't found, it looks for fallback files:
#     ./similarity_test.npy      (numpy array saved similarity matrix)
#     ./test_df.csv              (CSV exported test metadata)
#     ./ablation_results.csv     (CSV exported ablation table)
# - Each plot function will raise a clear error if inputs are missing.
#
# NOTE: Uses matplotlib only (no seaborn), one chart per figure (per your workshop submission needs).

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ---------- Helpers to resolve inputs ----------
def _resolve_similarity_and_df():
    # Try in-memory first
    g = globals()
    S_np = g.get("S_np", None)
    test_df = g.get("test_df", None)
    if S_np is not None and test_df is not None:
        return S_np, test_df

    # Fallback: try to load from disk
    sim_path = "similarity_test.npy"
    df_path = "test_df.csv"
    if os.path.exists(sim_path) and os.path.exists(df_path):
        S_np = np.load(sim_path)
        test_df = pd.read_csv(df_path)
        return S_np, test_df

    raise RuntimeError(
        "Missing inputs: expected variables S_np and test_df in memory, or files "
        "'similarity_test.npy' and 'test_df.csv' in the working directory."
    )

def _resolve_ablation_results():
    g = globals()
    res = g.get("res", None)
    if isinstance(res, pd.DataFrame) and {"remove_drug_name","T","margin","Grouped_R1"}.issubset(res.columns):
        return res.copy()

    # Fallback file
    ab_path = "ablation_results.csv"
    if os.path.exists(ab_path):
        res = pd.read_csv(ab_path)
        return res

    raise RuntimeError(
        "Missing ablation results: expected DataFrame `res` in memory with columns "
        "[remove_drug_name, T, margin, Grouped_R1] or an 'ablation_results.csv' file."
    )

# ---------- 1) Within-target similarity heatmap ----------
def plot_within_target_heatmap(target_id=None, max_items=40):
    """
    Draw an NxN cosine-similarity heatmap for a single target group.
    - target_id: the exact value from test_df['target_chembl_id']; if None, the largest group is chosen.
    - max_items: cap the subgroup size for readability.
    """
    S_np, test_df = _resolve_similarity_and_df()

    # Build index by target
    groups = test_df["target_chembl_id"].astype(str).fillna("UNK").values
    idx_by_group = {}
    for i, g in enumerate(groups):
        idx_by_group.setdefault(g, []).append(i)

    # Pick target
    if target_id is None:
        # choose largest group
        target_id = max(idx_by_group.keys(), key=lambda k: len(idx_by_group[k]))

    idxs = idx_by_group.get(str(target_id), [])
    if len(idxs) < 3:
        raise RuntimeError(f"Selected target group '{target_id}' has fewer than 3 examples. Pick another.")

    # Subsample to max_items
    idxs = idxs[:max_items]
    sub = S_np[np.ix_(idxs, idxs)]

    # Plot heatmap
    plt.figure(figsize=(6,5))
    im = plt.imshow(sub, vmin=np.min(sub), vmax=np.max(sub), aspect='auto')
    plt.colorbar(im, fraction=0.046, pad=0.04)
    plt.title(f"Within-target similarity (target={target_id}, n={len(idxs)})")
    plt.xlabel("Molecule index within group")
    plt.ylabel("Text index within group")
    plt.tight_layout()
    plt.savefig("within_target_heatmap.pdf")
    plt.show()

# ---------- 2) Top-k retrieval curves (CMC) ----------
def _recall_at_k(sim, k):
    gt = np.arange(sim.shape[0])
    # argsort descending, check if true index is in top-k
    topk = np.argpartition(-sim, kth=min(k, sim.shape[1]-1), axis=1)[:, :k]
    return np.mean([gt[i] in topk[i] for i in range(sim.shape[0])])

def _group_indices(meta_df, col="target_chembl_id", min_group=3):
    groups = meta_df[col].astype(str).fillna("UNK").values
    idx = {}
    for i, g in enumerate(groups):
        idx.setdefault(g, []).append(i)
    # filter small groups
    return {g:ix for g,ix in idx.items() if len(ix) >= min_group}

def plot_cmc_curves(ks=(1,3,5,10), min_group=3):
    """
    Plot cumulative match characteristic (Recall@k) curves for:
      - Global retrieval
      - Within-target retrieval (micro-averaged across groups with size >= min_group)
    """
    S_np, test_df = _resolve_similarity_and_df()

    # Global CMC
    global_vals = []
    for k in ks:
        global_vals.append(_recall_at_k(S_np, k))

    # Within-target CMC (micro-average)
    grouped = _group_indices(test_df, "target_chembl_id", min_group=min_group)
    within_vals = []
    for k in ks:
        hits = tot = 0
        for g, idxs in grouped.items():
            sub = S_np[np.ix_(idxs, idxs)]
            gt = np.arange(sub.shape[0])
            topk = np.argpartition(-sub, kth=min(k, sub.shape[1]-1), axis=1)[:, :k]
            hits += np.sum([gt[i] in topk[i] for i in range(sub.shape[0])])
            tot  += sub.shape[0]
        within_vals.append(hits / tot if tot else 0.0)

    # Plot
    plt.figure(figsize=(6,4.5))
    plt.plot(list(ks), global_vals, marker='o', label="Global")
    plt.plot(list(ks), within_vals, marker='s', label="Within-target")
    plt.title("Cumulative Match (Recall@k)")
    plt.xlabel("k")
    plt.ylabel("Recall@k")
    plt.grid(True, linestyle='--', linewidth=0.5)
    plt.legend()
    plt.tight_layout()
    plt.savefig("global_within_retrieval.pdf")
    plt.show()

# ---------- 3) Ablation bar chart for Grouped R@1 ----------
def plot_ablation_bar(top_n=8):
    """
    Plot top-N ablation settings ranked by Grouped_R1 as a bar chart.
    Expects columns: remove_drug_name, T, margin, Grouped_R1
    """
    res = _resolve_ablation_results()

    # Build short labels
    def _lbl(row):
        leak = "NoDrug" if row["remove_drug_name"] else "WithDrug"
        return f"{leak} | T={row['T']} | m={row['margin']}"

    res = res.copy()
    res["label"] = res.apply(_lbl, axis=1)
    res = res.sort_values("Grouped_R1", ascending=False).head(top_n)

    plt.figure(figsize=(8, 4.5))
    y = res["Grouped_R1"].values
    x = np.arange(len(res))
    plt.bar(x, y)
    plt.xticks(x, res["label"].tolist(), rotation=45, ha='right')
    plt.ylabel("Grouped Recall@1")
    plt.title("Ablation: Grouped R@1 by setting (top-N)")
    plt.tight_layout()
    plt.savefig("ablation_bar.pdf")
    plt.show()

print("Loaded plotting helpers:")
print(" - plot_within_target_heatmap(target_id=None, max_items=40)")
print(" - plot_cmc_curves(ks=(1,3,5,10), min_group=3)")
print(" - plot_ablation_bar(top_n=8)")
print("\nIf variables aren't in memory, save files as 'similarity_test.npy', 'test_df.csv', and 'ablation_results.csv' and re-run.")
```
